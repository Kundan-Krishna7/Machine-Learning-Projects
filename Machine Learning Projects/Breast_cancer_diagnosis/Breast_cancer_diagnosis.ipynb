**4.Data Preprocessing**

```python
#Seperate data into features and Target
X = df.drop(columns=["diagnosis"])
y = df["diagnosis"]
```

```python
# Train test split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
print("Dataset size", X.shape)
print("Train Dataset size", X_train.shape)
print("Test Dataset size",X_test.shape)
```

5.Baseline Model = Logistic Regression

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train_scaled, y_train)
```

```python
#train Accuracy
model.score(X_train_scaled, y_train)
```

6.Model Optimization

```python
#pipeline creation
pipeline=Pipeline(
    [
        ("scaler",StandardScaler()),
        ("pca",PCA()),
        ("model",SVC(kernel="rbf"))
    ]
)
```

```python
# Grid of Hyperparameters

param_grid = [
    {
        "pca__n_components": [10, 15, 20, 25, 30],

        "model__C": [0.1, 1, 10, 100, 1000],

        "model__gamma": [0.001, 0.01, 0.1, 1, "scale"]
    }
]
```

```python
#cross-validation setup(used inside grid search cv)
k = 5
skf= StratifiedKFold(n_splits=k, shuffle=True, random_state=RANDOM_STATE)
```

```python
# Hyperparameter Tuning

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring='recall',
    cv=skf,
    n_jobs=-1
)
```

```python
grid_search.fit(X_train,y_train)
```

```python
print("Hyperparameter Tuning Result:")
print("Best Params:", grid_search.best_params_)
print("Best Score:", grid_search.best_score_)
```

7.Train with best params

```python
best_pipeline = Pipeline(
    [
        ("scaler", StandardScaler()),

        ("PCA", PCA(n_components=10)),

        ("model", SVC(kernel="rbf", C=10, gamma="scale"))
    ]
)
```

```python
#train the pipeline
best_pipeline.fit(X_train, y_train)
```

8.Model Evaluation


```python
y_train_pred = best_pipeline.predict(X_train)
```

```python
train_acc = accuracy_score(y_train, y_train_pred)
print(f"Training Accuracy: {round(train_acc*100,2)}%")
```

```python
y_test_pred = best_pipeline.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_test_pred)
```

```python
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Pred Benign", "Pred Malignant"],
    yticklabels=["Actual Benign", "Actual Malignant"]
)
```

```python
print("Trainig Classification Report")
print(classification_report(y_train, y_train_pred))
```

```python
# Confusion Matrix

cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(4,3))

# heatmap
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Pred Benign", "Pred Malignant"],
    yticklabels=["Actual Benign", "Actual Malignant"]
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.title("Confusion Matrix Heatmap")

plt.show()
```

```python
# Test Data Prediction

y_test_pred = best_pipeline.predict(X_test)

# calculate accuracy
test_acc = accuracy_score(y_test, y_test_pred)

print(f"Test Accuracy: {round(test_acc*100, 2)}%")
```

9.Build a predictive system

```python
def predict_cancer(input_feature):

    input_df = pd.DataFrame(
        [input_feature],
        columns=X.columns
    )

    prediction = best_pipeline.predict(input_df)

    print("Model prediction:", prediction)

    if prediction[0] == "M":
        print("Diagnosis - Malignant 🔴")

    else:
        print("Diagnosis - Benign 🟢")
```
